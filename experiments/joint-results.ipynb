```python
from pathlib import Path
import json
import pandas as pd
RUN_DIR = Path.cwd()
plan = json.loads((RUN_DIR / "plan.json").read_text(encoding="utf-8"))
locked = json.loads((RUN_DIR / "selected.json").read_text(encoding="utf-8"))
print("Winner:", locked["config"])
print("Selection:", plan["selection_origins"], "Validation:", plan["holdout_origins"])
print(plan["monthly_policy"])
print(plan["warning"])
```

```python
for phase in ["selection", "holdout"]:
    path = RUN_DIR / (phase + "_summary.csv")
    if path.exists():
        print(phase)
        display(pd.read_csv(path))
        cases = pd.read_csv(RUN_DIR / (phase + "_cases.csv"))
        columns = ["composite", "marginal_contribution", "joint_contribution", "tail_contribution", "coverage90"]
        display(cases.groupby(["config", "cluster"])[columns].mean().groupby("config").mean())
        cells = pd.read_csv(RUN_DIR / (phase + "_cells.csv"))
        display(cells.groupby(["config", "asset", "horizon", "observation_steps"])[["marginal_crps", "baseline_crps"]].mean())
```
