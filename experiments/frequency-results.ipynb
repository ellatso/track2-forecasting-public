```python
from pathlib import Path
import json
import pandas as pd
RUN_DIR = Path.cwd()
plan = json.loads((RUN_DIR / "plan.json").read_text(encoding="utf-8"))
locked = json.loads((RUN_DIR / "selected.json").read_text(encoding="utf-8"))
print("Winners:", locked["winners"])
print("Selection:", plan["selection_origins"], "Validation:", plan["holdout_origins"])
print(plan["warning"])
```

```python
for frequency in ["daily", "monthly"]:
    for phase in ["selection", "holdout"]:
        prefix = phase + "_" + frequency
        path = RUN_DIR / (prefix + "_summary.csv")
        if path.exists():
            print(prefix)
            display(pd.read_csv(path))
            cases = pd.read_csv(RUN_DIR / (prefix + "_cases.csv"))
            display(cases.groupby("config")[["marginal_ratio", "joint_ratio", "tail_ratio", "coverage90"]].mean())
            cells = pd.read_csv(RUN_DIR / (prefix + "_cells.csv"))
            display(cells.groupby(["config", "asset", "horizon", "observation_steps"])[["marginal_crps", "baseline_crps"]].mean())
```
