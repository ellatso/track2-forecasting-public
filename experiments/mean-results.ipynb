## Executive summary (read this first)
Compare mean-only daily and monthly models separately. All dates are explored. Monthly historical release vintages are unavailable; monthly results are observation-index diagnostics. Local loss below 1 is better. No image or submission is changed.


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display
root = Path.cwd()
plan = json.loads((root/'plan.json').read_text(encoding='utf-8'))
display({k:plan[k] for k in ['draws','seeds','origins','warning']})
display(json.loads((root/'decision.json').read_text(encoding='utf-8')))
```

```python
for frequency in ['daily','monthly']:
    print(frequency.upper())
    for suffix in ['summary.csv','by_year.csv','by_family_seed.csv','components.csv','by_horizon.csv']:
        print(suffix)
        display(pd.read_csv(root/(frequency+'_'+suffix)))
    cases=pd.read_csv(root/(frequency+'_cases.csv'))
    display(cases.groupby(['config','cluster']).composite.mean().unstack(0))
coverage=json.loads((root/'coverage.json').read_text(encoding='utf-8'))
print('Exclusions:',{f:len(coverage[f]) for f in ['daily','monthly']})
print('Read paired.json intervals descriptively: baskets overlap.')
```
