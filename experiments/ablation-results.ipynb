## Executive summary (read this first)
Compare four fixed arms. Negative paired effects mean lower loss. These are previously explored historical diagnostics, not independent validation. Monthly models stay unchanged.

```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
plan = json.loads((root / 'plan.json').read_text(encoding='utf-8'))
print(plan['warning'])
display(pd.read_csv(root / 'diagnostic_summary.csv'))
display(pd.read_csv(root / 'diagnostic_by_year.csv').pivot(index='year', columns='config', values='local_relative_loss'))
```

```python
effects = pd.read_csv(root / 'effect_summary.csv')
display(effects)
print('Negative = lower loss. Interaction = both - variance-only - correlation-only + baseline.')
display(pd.read_csv(root / 'effect_by_year.csv').query('year == 2023'))
print(json.loads((root / 'paired_diagnostics.json').read_text(encoding='utf-8')))
```

```python
cases = pd.read_csv(root / 'diagnostic_cases.csv')
counts = cases.groupby(['group','origin','seed']).config.nunique()
assert (counts == 4).all()
assert not cases.duplicated(['group','origin','seed','config']).any()
worst = cases[cases.origin.str.startswith('2023')].groupby(['cluster','config']).composite.mean().unstack()
display(worst.sort_values('variance_and_correlation', ascending=False))
print('All four arms have matched cases. No submission created.')
```
